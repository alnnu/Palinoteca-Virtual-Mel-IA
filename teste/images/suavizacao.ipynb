```python
from skimage.io import imread, imshow
from skimage.filters import gaussian, prewitt, roberts, sobel, threshold_otsu, butterworth
from skimage.filters.rank import equalize, threshold
```

```python
img = imread("./puras/2025-01-29_17-09.jpg", as_gray=True)
imshow(img)
```

```python
gaussian_img = gaussian(img, sigma=5.0)

thresh = threshold_otsu(gaussian_img)

img_thresh = img > thresh 
imshow(img_thresh)
```

```python
print("Passa Alta")
img_butterworth = butterworth(img, 0.07, True, 7, channel_axis=-1)
imshow(img_butterworth)
```

```python
print("Passa Baixa")
img_butterworth = butterworth(img, 0.01, False, 4)
imshow(img_butterworth)
```

```python
img_gaussian = gaussian(img, sigma=0.4, mode="nearest", cval=0, preserve_range=False, truncate=4.0)
imshow(img_gaussian)
```

```python
img_prewitt = prewitt(img)
imshow(img_prewitt)
```

```python
img_roberts = roberts(img)
imshow(img_roberts)
```
